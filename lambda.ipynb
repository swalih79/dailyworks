```python
def students(name,age,mark):
    print(f"my name is {name}")
    print(f"i am {age}years old and my mark is {mark}")
    def depdets(schoolName,depname):
        print(schoolName)
        print(depname)
    depdets("a","b")
```

```python
students("swalih",21,50)

```

```python
students (mark=50, age=21, name="swalih")
```
